# Word2Vec with Thai alphabets Chapter 16 Workshop 4

## Load module

```python
from gensim.models import Word2Vec
```

## Create model

```python
url = "https://www.arts.chula.ac.th/ling/wp-content/uploads/TNCc5model.bin"

```

```python
import requests

response = requests.head(url)
if response.status_code == 200:
    print("URL is available.")
else:
    print(f"URL is not available. Status code: {response.status_code}")
```

```python
model = Word2Vec.load(url).wv
```

## Check vector

```python
len(model.index_to_key)  # Number of words in the vocabulary
```

```python
model.most_similar("สวัสดี", topn=5)  # Find words similar to "สวัสดี"
```

```python
model.similarity("สวัสดี", "สบายดี")  # Calculate similarity between two words
```

```python
model.doesnt_match(["สวัสดี", "สบายดี", "แมว", "หมา"])  # Find the odd one out in a list of words
```

## actual token

```python
from attacut import tokenize

text2 = "สาวสาวตากลมนั่งตากลมเขียนโปรแกรมประเทศไทย"
tokens = tokenize(text2)
print(tokens)  # Print the tokens from attacut for the new text
```

## Make unique

```python
import numpy as np
tokens_unique = np.unique(tokens)  # Get unique tokens
tokens_unique.tolist()  # Convert to list
```

## Create dictionary

```python
dict_th = {}
def create_thdict(tokens):
    dict_th = {}
    for idx, word in enumerate(tokens):
        dict_th[word] = idx + 1  # Assign an index starting from 1 
    return dict_th
tokens_unique.sort()  # Sort the tokens for consistency
dict_th = create_thdict(tokens_unique)  # Create a dictionary of tokens and their counts
dict_th

```

## Encoding

```python
word_enc = []
for wdi in tokens:
    word_enc.append(dict_th[wdi])  # Encode each word using the dictionary
print(word_enc)  # Print the encoded words
```
