## Grouping (zero shot classification)

### Load model

```python
zshot_classifier = pipeline(task="zero-shot-classification", model="MoritzLaurer/DeBERTa-v3-base-mnli-fever-anli")
```

### Define content

```python
content = "Now that you know what Asia Hotel Bangkok is like, your honest opinion will help others decide. Would you recommend it?"
candidate_label = ["vocation", "business", "political", "technology"]
```

### Prediction

```python
results = zshot_classifier(content, candidate_label)
```

```python
for i, val in enumerate(results['labels']):
    print(f"{val:>14} = {results['scores'][i]:.3f}")
```
