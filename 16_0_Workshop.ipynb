# Cleaning text (การทำความสะอาด)

```python
from bs4 import BeautifulSoup
html_str = '<td><a href="https://www.google.com">Click this link</a><br/><h1>Cybersecurity</h1></td>'
soup = BeautifulSoup(html_str)
print(html_str)
print(soup.get_text(separator="\n"))
```

# Bag-of-word (BOW)

```python
from sklearn.feature_extraction.text import CountVectorizer
from nltk.tokenize import word_tokenize
import nltk

# Download the required tokenizer model
nltk.download('punkt')


# Sentences
s1 = 'CNN is great.'
s2 = 'Python is a good language.'
s3 = 'You can learn CNN using Python.'

# List of sentences
corpus = [s1, s2, s3]

# Initialize the vectorizer
vectorizer = CountVectorizer()

# Fit and transform the corpus
X = vectorizer.fit_transform(corpus)

# Get feature names
features = vectorizer.get_feature_names_out()

# Convert to array or dataframe for display
print("Feature Names:", features)
print("Bag of Words Matrix:\n", X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Initialize the TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit and transform the corpus
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

# Get feature names
tfidf_features = tfidf_vectorizer.get_feature_names_out()

# Convert to pandas DataFrame for better visualization
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_features)

print("TF-IDF DataFrame:")
print(tfidf_df)
```
