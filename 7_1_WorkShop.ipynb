Mean Square Error [ Chapter 7 Workshop 1 ]

```python
from  sklearn.metrics import mean_squared_error
y_true = [3, -0.5, 2, 7]
y_pred = [2.5, 0.0, 2, 8]
print(mean_squared_error(y_true, y_pred))
```

Find Loss with Binary Cross-entropy [ Chapter 7 Workshop 2 ]

```python
import numpy as np
import tensorflow as tf

y_true = [0, 0, 1, 1]
y_pred = [0, 0, 1, 0]

y_true = np.array(y_true).astype('float')
y_pred = np.array(y_pred).astype('float')

bce = tf.keras.losses.BinaryCrossentropy()
# Calculate the loss with BinaryCrossentropy
loss = bce(y_true, y_pred).numpy()
print(loss.round(2))
```
