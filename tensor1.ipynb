```python
import tensorflow as tf
# Enable TensorFlow 1.x behavior

W1 = tf.ones((2,2))
W2 = tf.Variable(tf.zeros((2,2)), name="weights")
```

```python
# No need for a session in TensorFlow 2.x
print("W1:\n", W1.numpy())  # Directly evaluate the tensor

# Initialize variables (not necessary for tf.Variable in eager execution, but good practice)
W2.assign(tf.zeros((2, 2)))  # Reinitialize W2 to zeros (optional)

print("W2:\n", W2.numpy())  # Directly evaluate the variable
```
