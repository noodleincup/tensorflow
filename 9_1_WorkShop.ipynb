## Load module

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, AvgPool2D
from tensorflow.keras.models import Sequential
from tensorflow.image import rgb_to_grayscale
from tensorflow.keras.datasets.mnist import load_data
from PIL import ImageOps
import numpy as np
```

## Load image and prepare data

```python
# Load the MNIST dataset
(X_train, y_train), (X_test, y_test) = load_data()
test = X_test
train = X_train
```

```python
X_test = test
X_train = train
```

## Prepare data

```python
# use np.expand_dims
X_train = np.expand_dims(X_train, axis=-1)    # -1 means add the dimension at the last axis
X_test = np.expand_dims(X_test, axis=-1)      # -1 means add the dimension at the last axis
```

```python
X_train = np.pad(X_train, ((0, 0), (2, 2), (2, 2), (0, 0)), mode='constant')
X_test = np.pad(X_test, ((0, 0), (2, 2), (2, 2), (0, 0)), mode='constant')
```

```python
in_shape = X_train.shape[1:]
```

```python
class_names, y = np.unique(y_train, return_inverse=True)
print(f"Class names: {class_names}")
print(f"y: {y[:10]}")
print('y type', type(y))
```

```python
print('X_train shape: ',X_train.shape)
print('X_test shape: ',X_test.shape)
print('input shape:', in_shape)

print('y_test shape: ',y_test.shape)
print(y_train[:10][0])
```

## Nomalization

```python
## Normalize the image
X_train = X_train.astype(np.float32) / 255.0
X_test = X_test.astype(np.float32) / 255.0
```

```python
print('elements type:', X_train.dtype)
```

## Specific batch size

```python
number_sample = 10000
X_train = X_train[:number_sample]
y_train = y_train[:number_sample]
```

## Create model

```python
model = Sequential()

# Convo2D (filter = 6, kernel_size = (5, 5)) + Pooling #1
# Add a convolutional layer
model.add(Conv2D(6, (5, 5), activation='relu', input_shape=in_shape))   # C1
# Add a average pooling layer
model.add(AvgPool2D((2, 2)))                                            # S2

# Convo2D (filter = 16, kernel_size = (5, 5)) + Pooling #2
# Add a convolutional layer
model.add(Conv2D(16, (5, 5), activation='relu'))                        # C3
# Add a average pooling layer
model.add(AvgPool2D((2, 2)))                                            # S4

model.add(Conv2D(120, (5, 5), activation='relu'))                       # C5

# Flatten the output
model.add(Flatten())
# Add fully connected layer
model.add(Dense(84, activation='relu'))

# Add the output layer
model.add(Dense(10, activation='softmax'))

model.summary()
```

## Compile model

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

## Traning model

```python
history = model.fit(X_train, y_train, 
                    epochs=20, 
                    batch_size=128, 
                    validation_data=(X_test, y_test),
                    verbose=1)
```

# Evaluate model

```python
score = model.evaluate(X_test, y_test, verbose=0)
print('Test loss:', round(score[0], 3))
print('Test accuracy:', round(score[1], 3))
```
