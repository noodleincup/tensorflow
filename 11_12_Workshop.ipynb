## Load module

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
from PIL import Image, ImageOps
from IPython.display import display, Image
from tensorflow.keras.preprocessing.image import img_to_array, load_img, array_to_img
```

## Declare path

```python
model_path = '../models/oxford_segmentation.h5'
image_path = '../datasets/test_set/dog9056.jpg'
image_path_2 = '../datasets/test_set/cat9068.jpg'
```

## Load model

```python
model = load_model(model_path)
in_shape = (160, 160, 3)
```

## Load  image

```python
img = load_img(image_path_2, target_size=in_shape[:2])
img_test = np.expand_dims(img, axis=0)
```

## Predict image

```python
pred_mask = model.predict(img_test)[0]
mask = np.argmax(pred_mask, axis=-1)
mask = np.expand_dims(mask, axis=-1)
mask_img = ImageOps.autocontrast(array_to_img(mask))
```

## Show result

```python
display(img, mask_img)
display(array_to_img(pred_mask))
```
