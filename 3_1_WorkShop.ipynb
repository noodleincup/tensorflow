Procedure to use Scikit-learn for training model

```python
import numpy as np

x = [[0,0],
     [0,1],
     [1,0],
     [1,1]]

y = [0,0,0,1] # AND Gate
# y = [0,1,1,1] # OR Gate
# y = [0,1,1,0] # XOR Gate

X_train = np.array(x)
y_train = np.array(y)
```

```python
#  import Scikit-learn library
from sklearn.linear_model import Perceptron

# Define the model
model = Perceptron(tol=1e-3, random_state=0, verbose=True) # Create Neuron 1 hidden layer

# Train the model
model.fit(X_train, y_train)

print('score: ', model.score(X_train, y_train))
print('predict: ', model.predict(X_train))
print('target: ', y_train)
```

```python
# check weights parameter
print('weights: ', model.coef_[0])
print('w0 (bias): ', model.intercept_[0])
```

Use model that have multi-hidden layers

```python
#  import Scikit-learn library
from sklearn.neural_network import MLPClassifier

# Define the model
model = MLPClassifier(hidden_layer_sizes=4 , random_state=1, verbose=True, max_iter=900) # Create NN (number of hidden layers = 3)
model.fit(X_train, y_train)

print('score: ', model.score(X_train, y_train))
print('predict: ', model.predict(X_train))
print('target: ', y_train)
```

```python
# Check MLP weights parameter shape
[coef.shape for coef in model.coefs_]

# [(2, 4), (4, 1)] : 2 input, 1 hidden layer with 4 nodes, output layer with 1 node
```

```python
print(model.coefs_)
print("----------------------")
# print(model.coefs_[1])

# Check MLP bias parameter

print(model.intercepts_)
# print(model.intercepts_[1])
```

Visualization

```python
import matplotlib.pyplot as plt
# Define colors for each class
colors = ['red' if y == 0 else 'blue' for y in y_train]

# Scatter plot
plt.figure(figsize=(5,5))
plt.scatter(X_train[:,0], X_train[:,1], c=colors, edgecolors='black', s=100)

# Labels and title
plt.xlabel('X1')
plt.ylabel('X2')
plt.title('Visualization of XOR Gate')

# Show grid
plt.grid(True)
plt.xticks([0,1])
plt.yticks([0,1])

# Show plot
plt.show()
```
