Classification with 2 class

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("D:\\Programming_File\\Test\\fruit2class2.csv")
df.sample(7, random_state=0)
```

```python
# check number of classes
df.label.unique()

# Data Visualization
sns.scatterplot(x='m', y='d', data=df, hue='label', style='label', s=140)
plt.show()
```

Prepare data before traning

```python
X = df.drop('label', axis=1)
y = df.label

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

X_train = sc.fit_transform(X)   # X for train
y_train = y                     # y for train
df.sample(7, random_state=0)    
```

```python
# Nomalization
df_sc = pd.DataFrame(X_train, columns=['d','m'])
df_sc['label'] = y_train

df_sc.sample(10, random_state=0)
```

```python
# Data Visualization
sns.scatterplot(x='m', y='d', data=df_sc, hue='label', style='label', s=140)
plt.show()
```

Create Neural Network Model

```python
from sklearn.neural_network import MLPClassifier

hidden =  150 # Default: 100 
iter = 200 # Default: 200

# Define model

# Case 1 Defalut
# model = MLPClassifier(random_state=1)
# Case 2 Increase Iteration
# model = MLPClassifier(max_iter=2000 ,random_state=1)
# Case 3 Increase hidden layer node
# model = MLPClassifier(hidden_layer_sizes= 150, max_iter=200 ,random_state=1)
# Case 5 Increase Iteration and Decrease hidden layer node
model = MLPClassifier(hidden_layer_sizes= 10, max_iter=2000 ,random_state=1)

# Train the model
model.fit(X_train, y_train)

print('score:', round(model.score(X_train,y_train),4))

print('annual y:', np.array(y))
print('predicted: ', model.predict(X_train))
```

Report the results

```python
from sklearn.metrics import classification_report, confusion_matrix
# prediction
y_predict = model.predict(X_train)

# report
print('score:', round(model.score(X_train,y_train),4))
print(classification_report(y_predict, y_train))
print(confusion_matrix(y_train,y_predict))
```

Check area of model classification

```python
from mlxtend.plotting import plot_decision_regions

plt.figure(figsize=(4.5,3.5))
plot_decision_regions(X_train, np.array(y_train), clf=model, legend=2)

plt.title('MLP: 2 Classes (hidden layer: {})'.format(hidden))
plt.xticks([])
plt.yticks([])
plt.xlabel('m (gram)')
plt.ylabel('d (diameter)')
plt.show()
```

show model construction

```python
[coef.shape for coef in model.coefs_] # 2 input, 1 hidden layer 100 nodes, 1 output
```

```python
print(model.coefs_) # (2 rows, 100 columns)
```

Plot Loss Curve 

```python
plt.figure(figsize=(5,3))
plt.title('Loss Curve')
plt.xlabel('iteration')
plt.plot(model.loss_curve_)
plt.show()
```

Create dataset 1

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Generate data for Grapes (label = 0)
grape_mass = np.random.normal(6, 2, 150).clip(3, 12)  # Mean = 6g, SD = 2g
grape_diameter = np.random.normal(16, 2, 150).clip(12, 22)  # Mean = 16mm, SD = 2mm

# Generate data for Lemons (label = 1)
lemon_mass = np.random.normal(120, 20, 50).clip(80, 160)  # Mean = 120g, SD = 20g
lemon_diameter = np.random.normal(55, 10, 50).clip(40, 80)  # Mean = 55mm, SD = 10mm

# Add some "confusing" Grapes (bigger than normal) and Lemons (smaller than normal)
grape_mass_extra = np.random.normal(30, 10, 15).clip(15, 50)  # Some big Grapes
grape_diameter_extra = np.random.normal(40, 5, 15).clip(30, 50)

lemon_mass_extra = np.random.normal(50, 10, 10).clip(30, 80)  # Some small Lemons
lemon_diameter_extra = np.random.normal(35, 5, 10).clip(25, 45)

# Combine data
mass = np.concatenate((grape_mass, grape_mass_extra, lemon_mass, lemon_mass_extra))
diameter = np.concatenate((grape_diameter, grape_diameter_extra, lemon_diameter, lemon_diameter_extra))
labels = [0] * (150 + 15) + [1] * (50 + 10)

# Create DataFrame
df = pd.DataFrame({"m": mass, "d": diameter, "label": labels})
# Save as CSV
df.to_csv("D:\\Programming_File\\Test\\fruit2class.csv", index=False)

print("CSV file 'fruit_data.csv' has been created.")
```
