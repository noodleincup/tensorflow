# Word2Vec Model from Google Chapter 16 Workshop 1

## Load module

```python
import gensim
gensim.__version__
```

## Load model

```python
import gensim.downloader as api
from gensim.models import KeyedVectors
import os
# Load the pre-trained Word2Vec model
# model = api.load("word2vec-google-news-300")
model_path = '../models/GoogleNews-vectors-negative300/GoogleNews-vectors-negative300.bin'
if not os.path.exists(model_path):
    print(f"Model file not found at {model_path}. Please download it from the Gensim data repository.")
else:
    model = KeyedVectors.load_word2vec_format(model_path, binary=True)
    print('Loaded model successfully.')
```

## Check model

### vector size and key numbers

```python
print(f'Vector size: {model.vector_size}')
print(f'Key numbers: {len(model.key_to_index)}')
```

### 'dog' vector

```python
print('Dog vector shape:', model['dog'].shape)
print('Dog vector')
print(model['dog'])
```

### the first 10 words of model

```python
for index, word in enumerate(model.index_to_key[:10]):
    print(f'{index}: {word}')
```

### unexisted word

```python
try:
    vec = model['Deep learning']
except KeyError as e:
    print(f"KeyError: {e}. The word 'Deep learn' is not in the vocabulary. Use 'deep_learn' instead.")
```

### simarity

```python
print(f'"big" and "huge" similarity: {model.similarity("big", "huge"):.4f}')
print(f'"big" and "small" similarity: {model.similarity("big", "small"):.4f}')
print(f'top 8 similar words to "car": {model.most_similar("car", topn=8)}')
```

```python
pairs = [('car', 'automobile'), 
         ('car', 'minivan'), 
         ('car', 'bicycle'), 
         ('car', 'airplane'),
         ('car', 'apple'),
         ('car', 'man')]

for word1, word2 in pairs:
    try:
        similarity = model.similarity(word1, word2)
        print('%r\t%r\t%.4f' % (word1, word2, similarity))
    except KeyError as e:
        print(f"KeyError: {e}. One of the words '{word1}' or '{word2}' is not in the vocabulary.")
```

```python
model.most_similar(positive=['car', 'minivan'], topn=5)
```

### Odd one out

```python
words = ['movie', 'song', 'book', 'music']
print(model.doesnt_match(words))
```

### simarity by vector

```python
vec = model['king'] - model['man'] + model['women']
print(f'simarity: {model.most_similar(vec, topn=5)}') # model.most_similar(positive=['king', 'woman'], negative=['man'], topn=5)
print(f'simarity: {model.similar_by_vector(vec, topn=5)}')
```
