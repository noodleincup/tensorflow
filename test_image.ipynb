# import module

```python
import numpy as np
from tensorflow.image import rgb_to_grayscale
from PIL import Image, ImageOps
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import matplotlib.pyplot as plt
```

# define image file

```python
img_file = "D:\Project\Learning\TensorFlow_Files\image\digit\digit1.png"
```

## Load image method 1

```python
# Load the image with RGB and resize
img1 = load_img(img_file, target_size=(28, 28))
# Invert the image colors
img1 = ImageOps.invert(img1)  # Invert the image colors
# Tranform to numpy array
img1_arr = img_to_array(img1)
img1_arr = rgb_to_grayscale(img1_arr)
```

## Load image method 2

```python
# Step 1: Open image with PIL
img2 = Image.open(img_file)

# Step 2: Convert to RGB (to match load_img behavior)
img2 = img2.convert("RGB")

# Step 3: Resize (like target_size in load_img)
img2 = img2.resize((28, 28), Image.Resampling.LANCZOS)

# Step 4: Invert colors in RGB mode
img2 = ImageOps.invert(img2)

# Step 5: Convert to NumPy array (same as img_to_array)
img2_arr = img_to_array(img2)

# Step 6: Convert RGB to grayscale using TensorFlow's method
img2_arr = rgb_to_grayscale(img2_arr)
```

# Debug result

```python
print('img1 type',type(img1))
print('img2 type',type(img2))
print('img1 == img2', img1 == img2)

print(img1_arr.shape)
print(img2_arr.shape)

print('max 1: ', np.max(img1_arr))
print('max 2: ',np.max(img2_arr))
print('min 1: ',np.min(img1_arr))
print('min 2: ',np.min(img2_arr))

check_el = img1_arr == img2_arr
check_all = np.all(check_el)
print('check_all', check_all)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img1)
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(img2)
plt.axis('off')
plt.show()

# print(img1_arr)
# print(img2_arr)
```
