Dot Product for sum calculation

```python
import numpy as np
w0 = -0.5
w = [.5, .5]
x = [[1],[1]]

x = np.array(x)
w = np.array(w)

sum = w0 + np.dot(w, x)

print(sum)
```

Thershold

```python
if (sum > 0):
    print(1)
else:
    print(0)
```
